## Теоретический расчет

Условия:
* Стена - общая область исходов
* Мишень — это благоприятная область, куда мы хотим попасть.

Изветсно, что $S_{мишень} = \frac{1}{2} * S_{стены}$.

По формуле геом. вероятности:
$P(A) = \frac{S_{благоприятная}}{S_{общая}} = \frac{S_{мишени}}{S_{стены}} = \frac{1}{2} = 0,5$ 


```python
import matplotlib.pyplot as plt
from matplotlib.patches import Rectangle, Circle
import numpy as np
```

```python
def darts(size):
    x0, y0, R = 6, 6, 5
    side = np.sqrt(2* np.pi * R**2)
    fig, ax = plt.subplots(figsize=(6, 6))
    ax = plt.gca()
    ax.add_patch(Rectangle((0, 0), side, side))
    ax.add_patch(Circle((x0, y0), R, color='r'))

    X = np.random.uniform(0, side, size)
    Y = np.random.uniform(0, side, size)
    plt.scatter(X, Y, marker='x', color='black')

    inside = len([i for i in range(size) if (X[i] - x0)**2 + (Y[i] - y0)**2 < R**2])
    prob = inside / size

    plt.show()
    print(f'В красный круг попало {inside} точек из {size}. Вероятность попадания: {prob:.4f}')
    

```

```python
darts(1000)
```

### Задача: «Сигнал микронаушника на экзамене»

Экзаменационная аудитория имеет прямоугольную форму размером **12 × 8 м**. В центре аудитории находится устройство, передающее сигнал на микронаушник. Устойчивый сигнал принимается на расстоянии не более **4 м** от устройства.

Студент случайным образом занимает место в аудитории.

Найти вероятность того, что выбранное место окажется в зоне устойчивого сигнала микронаушника.
### Теоретический расчет:

$S_{ауд.} = 12 * 8 = 98$
$S_{wi-fi} = π * 4^{2} = 16*π$

Пусть событие А - выбранное место окажется в зоне устойчивого сигнала микронаушника.

$P(A) = \frac{16*π}{96} ​≈0.524$


```python
import matplotlib.pyplot as plt
import numpy as np


def random_points(n):
    # Генерируем случайные координаты точек
    x = np.random.uniform(0, 12, n)
    y = np.random.uniform(0, 8, n)

    # Проверяем попадание каждой точки в круг
    hits = (x - 6) ** 2 + (y - 4) ** 2 <= 4 ** 2

    # Считаем экспериментальную вероятность
    probability = np.sum(hits) / n

    return x, y, probability


fig, ax = plt.subplots()

# Аудитория 12 × 8
rectangle = plt.Rectangle(
    (0, 0),
    12,
    8,
    facecolor='lightblue',
    edgecolor='black',
    linewidth=2
)

ax.add_patch(rectangle)

# Зона устойчивого сигнала
circle = plt.Circle(
    (6, 4),
    4,
    facecolor='lightgreen',
    edgecolor='green',
    linewidth=2
)

ax.add_patch(circle)

# Генерируем 1000 случайных точек
x, y, probability = random_points(1000)

# Рисуем точки крестиками
ax.scatter(x, y, marker='x')

# Границы графика
ax.set_xlim(-1, 13)
ax.set_ylim(-1, 9)

# Одинаковый масштаб осей
ax.set_aspect('equal')

# Сетка
plt.grid()

# Вывод вероятности


plt.show()
```

```python
x, y, probability = random_points(1000)
print("Экспериментальная вероятность:", probability)
```
